```python
from bs4 import BeautifulSoup
import requests
```

```python
# Define the URL and download the page
url = "https://www.courtlistener.com/docket/69609277/penguin-random-house-llc-v-labrador/"
headers = {
    "User-Agent": "Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/122.0.0.0 Safari/537.36"
}
response = requests.get(url, headers=headers)
html = response.text
```

```python
# Parse the HTML using BeautifulSoup
soup = BeautifulSoup(html, "html.parser")
print(soup.title.string)
print(url)
```

```python
# Find all divs with an id starting with 'entry-'
entry_ids = [div.get("id") for div in soup.find_all("div", id=True) if div.get("id").startswith("entry-")]

# Extract the numbers from the entry-ids
entry_numbers = [int(entry_id.split('-')[1]) for entry_id in entry_ids]

# Find the highest entry number
max_entry = max(entry_numbers)

# Print the first entry (entry-1)
first_entry = soup.find("div", id="entry-1")
if first_entry:
    # Extract the <a> tags within the div and print the text and link for each
    for a_tag in first_entry.find_all("a", href=True):
        link_text = a_tag.text.strip()  # Get the text of the hyperlink
        link_url = a_tag.get("href")  # Get the URL of the hyperlink
        print(f"{link_text} > {link_url}")

```

```python
# Print the last entry (entry-max_entry)
last_entry = soup.find("div", id=f"entry-{max_entry}")
if last_entry:
    # Extract the <a> tags within the div and print the text and link for each
    for a_tag in last_entry.find_all("a", href=True):
        link_text = a_tag.text.strip()  # Get the text of the hyperlink
        link_url = a_tag.get("href")  # Get the URL of the hyperlink
        print(f"{link_text} > {link_url}")
```
